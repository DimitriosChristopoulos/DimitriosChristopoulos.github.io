```python
taj_img = np.array(cv.imread('2.1/images/taj.jpg'))
agia_img = np.array(cv.imread('2.1/images/agia_sophia.jpg'))
```

```python
blurred_taj = taj_img.copy()
for channel in range(3):
    blurred_taj[:, :, channel] = scipy.signal.convolve2d(blurred_taj[:, :, channel], (cv.getGaussianKernel(ksize=9, sigma=0) @ cv.getGaussianKernel(ksize=9, sigma=0).T), mode='same')

save_img(blurred_taj, '2.1/results/blurred_taj.jpg')
```

```python
blurred_agia = agia_img.copy()
for channel in range(3):
    blurred_agia[:, :, channel] = scipy.signal.convolve2d(blurred_agia[:, :, channel], (cv.getGaussianKernel(ksize=9, sigma=0) @ cv.getGaussianKernel(ksize=9, sigma=0).T), mode='same')

save_img(blurred_agia, '2.1/results/blurred_agia.jpg')
```

```python
sharpened_taj = unsharp_mask_filter(blurred_taj)
sharpened_agia = unsharp_mask_filter(blurred_agia)
```

```python
save_img(sharpened_taj, '2.1/results/sharpened_taj.jpg')
save_img(sharpened_agia, '2.1/results/sharpened_agia.jpg')
```
